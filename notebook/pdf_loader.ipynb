```python
import os

from langchain_community.document_loaders import PyPDFLoader , PyMuPDFLoader

from langchain_text_splitters import RecursiveCharacterTextSplitter
from pathlib import Path
```

```python
### read all pdfs

def process_pdf_files(pdf_directory):
    all_documents = []
    pdf_directory = Path(pdf_directory)
    pdf_files = list(pdf_directory.glob("*.pdf"))
    print(f"Found {len(pdf_files)} PDF files in the directory: {pdf_directory}")
    

    for pdf_file in pdf_files:
        # Load the PDF file
        print(f"Processing PDF file: {pdf_file.name}")
        try:
            loader = PyMuPDFLoader(str(pdf_file))
            documents = loader.load()

            for doc in documents:
                doc.metadata["source"] = pdf_file.name  # Add the source metadata
                doc.metadata["file_type"] = "pdf"  # Add the file type metadata

            all_documents.extend(documents)
            print(f" loaded {len(documents)} pages")
        except Exception as e:
            print(f"Error processing {pdf_file.name}: {e}")

    print(f"Total documents after splitting: {len(all_documents)}")

    return all_documents

all_documents = process_pdf_files("../data/pdf")


```

```python
### text splittng

def text_chunking(documents, chunk_size=1000, chunk_overlap=200):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
        separators=["\n\n", "\n", " ", ""]
    )

    split_docs= text_splitter.split_documents(documents)
    print(f"{(len(documents))} --> Before \nTotal documents after splitting: {len(split_docs)}")
    if split_docs:
        print(f"First document after splitting: {split_docs[0].page_content[:200]}...")  # Print first 200 characters of the first document
        print(f"Metadata: {split_docs[0].metadata}")  # Print metadata of the first document

    return split_docs
```

```python
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb
from chromadb.config import Settings
import uuid
from typing import List, Dict , Any , Tuple
from sklearn.metrics.pairwise import cosine_similarity
```

```python
class EmbeddingManager:
    """handles document embedding"""

    def __init__(self, model_name: str = "all-MiniLM-L6-v2"):
        """initiazlize the embedding model"""
        self.model_name = model_name
        self.model = None

        self._load_model()

    def _load_model(self):
        """Load the embedding model"""
        try:
            self.model = SentenceTransformer(self.model_name)
            print(f"Loaded embedding model: {self.model_name}")
            print(f"Model details: {self.model.get_embedding_dimension()} dimensions")
        except Exception as e:
            print(f"Error loading embedding model {self.model_name}: {e}")
            raise

    def generate_embeddings(self, texts: List[str]) -> np.ndarray:
        """Generate embeddings for a list of texts"""
        if not self.model:
            raise ValueError("Embedding model is not loaded.")
        try:
            embeddings = self.model.encode(texts, convert_to_numpy=True,normalize_embeddings=True)
            print(f"Generated embeddings for {len(texts)} texts. Embedding shape: {embeddings.shape}")
            return embeddings
        except Exception as e:
            print(f"Error generating embeddings: {e}")
            raise



    ### intizialize the embedding manager

embedding_manager = EmbeddingManager(model_name="all-MiniLM-L6-v2")
    
```

```python
class VectorStore:
    def __init__(self, collection_name: str = "pdf_documents_v5", persist_directory: str = "../data/vector_store"):
        """Initialize the vector store"""
        self.collection_name = collection_name
        self.persist_directory = persist_directory
        self.client = None
        self.collection = None

        self._initialize_vector_store()

    def _initialize_vector_store(self):

        try:
            os.makedirs(self.persist_directory, exist_ok=True)
            self.client = chromadb.PersistentClient(path=self.persist_directory)

            self.collection = self.client.get_or_create_collection(name=self.collection_name,metadata={"hnsw:space": "cosine","description": "PDF document embeddings for RAG"})

            print(f"Initialized vector store with collection: {self.collection_name} at {self.persist_directory}")
        except Exception as e:
            print(f"Error initializing vector store: {e}")
            raise   

    def add_documents(self, documents: List[Dict[str, Any]], embeddings: np.ndarray):
        """Add documents and their embeddings to the vector store"""
        if len(documents) != embeddings.shape[0]:
            raise ValueError("Number of documents and embeddings must match.")

        print(f"adding {len(documents)} documents to the vector store")

        ids = []
        metadatas=[]
        documents_text=[]
        embeddings_list = []

        for i, (doc, embedding) in enumerate(zip(documents, embeddings)):
            doc_id = str(uuid.uuid4())
            ids.append(doc_id)

            #prepare metadata and document text
            metadata = dict(doc.metadata)  # Ensure metadata is a dictionary
            metadata['doc_index'] = i
            metadata['content-length'] = len(doc.page_content)
            metadatas.append(metadata)

            #document text
            documents_text.append(doc.page_content)

            #embedding
            embeddings_list.append(embedding.tolist())


            #add to collection
        try:
            self.collection.add(
                ids=ids,
                embeddings=embeddings_list,
                metadatas=metadatas,
                documents=documents_text
                )
            print(f"Added document {i+1}/{len(documents)} with ID: {doc_id}")
        except Exception as e:
            print(f"Error adding document {i+1} to vector store: {e}")
            raise



vector_store = VectorStore(collection_name="pdf_documents_v5", persist_directory="../data/vector_store")
vector_store
```

```python


class RAGRetriever:
    """Handles document retrieval from vector store"""

    def __init__(self, vector_store: VectorStore, embedding_manager: EmbeddingManager):
        self.vector_store = vector_store
        self.embedding_manager = embedding_manager

    def retrieve(self, query: str, top_k: int = 5, score_threshold: float = 0.0) -> List[Dict[str, Any]]:
        """Retrieve relevant documents based on the query"""
        print(f"Retrieving for query: '{query}'")
        print(f"Top K: {top_k}, Score Threshold: {score_threshold}")

        try:
            # 1. Generate embedding for the query
            query_embedding = self.embedding_manager.generate_embeddings([query])[0]

            # 2. Query vector store
            results = self.vector_store.collection.query(
                query_embeddings=[query_embedding.tolist()],
                n_results=top_k,
            )

            retrieved_docs = []

            # 3. Process results safely inside the IF block
            if results.get('documents') and results['documents'][0]:
                documents = results['documents'][0]
                metadatas = results['metadatas'][0]
                distances = results['distances'][0]
                ids = results['ids'][0]

                for i, (doc_id, doc, metadata, distance) in enumerate(zip(ids, documents, metadatas, distances)):
                    similarity_score = 1 - (distance / 2) # Convert distance to similarity score

                    if similarity_score >= score_threshold:
                        retrieved_docs.append({
                            "id": doc_id,
                            "document": doc,
                            "metadata": metadata,
                            "similarity_score": round(similarity_score, 4),
                            "distance": round(distance, 4),
                            "rank": i + 1
                        })

            # 4. Print summary ONCE outside the loop
            if retrieved_docs:
                print(f"Successfully retrieved {len(retrieved_docs)} document(s) matching threshold >= {score_threshold}")
            else:
                print(f"No documents found meeting similarity threshold >= {score_threshold}")

            return retrieved_docs

        except Exception as e:
            print(f"Error during retrieval: {e}")
            return []

# Initialize the retriever
rag_retriever = RAGRetriever(vector_store=vector_store, embedding_manager=embedding_manager)


            
    

            





```

ADVANCED RAG PIPELINE

```python
def rag_advanced(
    query, retriever, llm, top_k=5, return_context=False, min_score=0.40
):
  """Advanced RAG function with safety threshold floor and fallback logic."""
  # 1. Force a minimum safety floor so score=0 doesn't bypass fallback
  effective_threshold = max(min_score, 0.35)

  # 2. Retrieve documents
  raw_results = (
      retriever.retrieve(
          query, top_k=top_k, score_threshold=effective_threshold
      )
      or []
  )

  # 3. Hard filter out low-scoring garbage matches
  results = [
      doc
      for doc in raw_results
      if doc.get("similarity_score", 0) >= effective_threshold
  ]

  if results:
    context = "\n\n".join([
        f"Document {doc['rank']} (Score: {doc['similarity_score']}):\n{doc['document']}"
        for doc in results
    ])
    sources = [
        {
            "rank": doc["rank"],
            "score": doc["similarity_score"],
            "page": doc["metadata"].get(
                "page", doc["metadata"].get("doc_index", "unknown")
            ),
            "preview": doc["document"][:300] + "...",
        }
        for doc in results
    ]
    confidence_score = max(doc["similarity_score"] for doc in results)

    # Grounded prompt with emergency fallback clause
    prompt = f"""You are a helpful AI agriculture and food safety assistant.
1. If the provided context is RELEVANT to the query, answer the query strictly using the context.
2. If the context is UNRELATED to the query (e.g. asking about scuba diving when context is about OOPs or SSOPs), ignore the context completely and answer using your general knowledge. Start your answer with: "[Note: Database context was unrelated. Answering via general knowledge.] try to keep the answer below 2048 tokens"

Context:
{context}

Query: {query}

Answer:"""

  else:
    context = ""
    sources = []
    confidence_score = 0.0

    prompt = f"""Answer the following question using your general pre-trained knowledge.

You MUST start your response with this exact disclaimer:
"[Note: No relevant documents were found in the database. This response is auto-generated using general knowledge.]"

Query: {query}

Answer:"""

  # 4. Execute LLM call
  response = llm.invoke(prompt)
  answer_text = (
      response.content if hasattr(response, "content") else str(response)
  )

  output = {
      "answer": answer_text,
      "sources": sources,
      "confidence_score": confidence_score,
  }

  return output if return_context else answer_text
```

```python
import os
import asyncio
from typing import TypedDict, List, Literal
from pydantic import BaseModel, Field

from langgraph.graph import StateGraph, END
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq

# ==========================================
# 1. STATE & SCHEMA DEFINITIONS
# ==========================================
class AgentState(TypedDict):
    question: str
    documents: List[Document]
    answer: str
    intent: str  # "domain_query" or "general_query"

class IntentClassification(BaseModel):
    """Classifies user intent."""
    intent: Literal["domain_query", "general_query"] = Field(
        description="Choose 'domain_query' if the question is about compliance, safety, logistics, cold storage, transit, or domain knowledge. "
                    "Choose 'general_query' for greetings, off-topic questions, general chitchat, or programming/general knowledge."
    )




# ==========================================
# 2. ROUTER & PIPELINE NODES
# ==========================================

async def classify_intent_node(state: AgentState) -> dict:
    """Classifies whether the query needs RAG or direct LLM answer."""
    question = state["question"]
    classifier_llm = llm.with_structured_output(IntentClassification)
    
    prompt = ChatPromptTemplate.from_messages([
        ("system", "Analyze the user prompt. Classify whether it requires searching a specialized domain knowledge base (domain_query) "
                   "or if it is a general greeting, off-topic question, or general request (general_query)."),
        ("human", "{question}")
    ])
    
    try:
        res = await (prompt | classifier_llm).ainvoke({"question": question})
        intent = res.intent
    except Exception:
        intent = "domain_query"  # Default fallback if classification fails
        
    return {"intent": intent}


def route_intent(state: AgentState) -> str:
    """Conditional Edge Router."""
    if state["intent"] == "domain_query":
        return "retrieve"
    return "direct_answer"


async def retrieve_documents(state: AgentState) -> dict:
    """Fetches relevant context from vector store."""
    question = state["question"]
    
    if hasattr(rag_retriever, "ainvoke"):
        retrieved_docs = await rag_retriever.ainvoke(question)
    else:
        loop = asyncio.get_running_loop()
        retrieved_docs = await loop.run_in_executor(
            None, lambda: rag_retriever.retrieve(question, top_k=3)
        )
    return {"documents": retrieved_docs}


async def generate_rag_answer(state: AgentState) -> dict:
    """Answers domain queries using retrieved vector DB context."""
    question = state["question"]
    docs = state["documents"]

    if not docs:
        return {"answer": "I do not have sufficient information in my compliance database to answer this question."}

    context = "\n\n".join([d.page_content if hasattr(d, 'page_content') else str(d) for d in docs])

    prompt = ChatPromptTemplate.from_messages([
        ("system", "Answer the question strictly using ONLY the provided context. If the context does not contain the answer, "
                   "clearly state that you do not have sufficient information in your database."),
        ("human", "Context:\n{context}\n\nQuestion: {question}")
    ])

    response = await (prompt | llm).ainvoke({"context": context, "question": question})
    return {"answer": response.content}


async def generate_direct_answer(state: AgentState) -> dict:
    """Answers general, off-topic, or conversational queries directly."""
    question = state["question"]
    
    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a polite agricultural food safety AI  assistant. Respond to general greetings, user inquiries, or general questions helpful and concisely. "
                   "If asked off-topic questions outside your scope, answer naturally while briefly introducing your role as a Agricultural Food Safety Assistant."),
        ("human", "{question}")
    ])

    response = await (prompt | llm).ainvoke({"question": question})
    return {"answer": response.content}


# ==========================================
# 3. GRAPH COMPOSITION
# ==========================================

workflow = StateGraph(AgentState)

workflow.add_node("classify_intent", classify_intent_node)
workflow.add_node("retrieve", retrieve_documents)
workflow.add_node("generate_rag", generate_rag_answer)
workflow.add_node("direct_answer", generate_direct_answer)

workflow.set_entry_point("classify_intent")

workflow.add_conditional_edges(
    "classify_intent",
    route_intent,
    {
        "retrieve": "retrieve",
        "direct_answer": "direct_answer"
    }
)

workflow.add_edge("retrieve", "generate_rag")
workflow.add_edge("generate_rag", END)
workflow.add_edge("direct_answer", END)

app = workflow.compile()


# ==========================================
# 4. ASYNC ENTRYPOINT FOR GRADIO
# ==========================================

async def ask_question_async(question: str) -> str:
    initial_state = {"question": question, "documents": [], "answer": "", "intent": ""}
    result = await app.ainvoke(initial_state)
    return result["answer"]
```
