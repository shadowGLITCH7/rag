```python
import os

os.makedirs("../data/text_files", exist_ok=True)

```

```python
sample_text = {"../data/text_files/python.txt": """Our team reviewed customer feedback from the previous quarter and identified several recurring requests for faster service, more transparent support, and clearer product updates. Customers value trustworthy communication, consistent delivery, and reliable guidance when choosing solutions. The workshop focused on improving onboarding, strengthening documentation, and creating practical examples for everyday workflows. These changes should help users feel confident, supported, and prepared to succeed. Encouraging collaboration across departments will also reduce delays and improve " \
"decision making for " \
"future " \
"releases, " \
"creating a stronger foundation for both customer satisfaction and long term growth." \
" The organization plans to measure progress and share outcomes monthly."""}

for filepath,content in sample_text.items():
    with open(filepath, "w" , encoding="utf-8") as f:
        f.write(content)

print("sample_text.txt file created successfully.")
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python.txt",encoding="utf-8")
document=loader.load()
print(document)



```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader("../data/text_files", 
                             glob="*.txt", 
                             loader_cls=TextLoader, 
                             loader_kwargs={"encoding": "utf-8"})
directory_documents = dir_loader.load()
print(directory_documents)
```

```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
dir_loader = DirectoryLoader("../data/pdf", 
                             glob="*.pdf", 
                             loader_cls=PyMuPDFLoader, 
                             )
directory_documents = dir_loader.load()
print(directory_documents)
```
