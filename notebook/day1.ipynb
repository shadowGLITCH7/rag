```python
import pandas as pd

url = "https://huggingface.co/datasets/McAuley-Lab/Amazon-Reviews-2023/resolve/main/raw/meta_categories/meta_Appliances.jsonl"

# Read directly into a DataFrame
df = pd.read_json(url, lines=True)

# Convert to a list of dicts (matches dataset[i] structure)
dataset = df.to_dict(orient="records")

# Test on the first item
print(dataset[0]["title"])
```

```python
from tqdm import tqdm
from traitlets import Any


max_price=0
max_item=None

for dp in tqdm(dataset):
    try:
        price = float(dp["price"])
        if price > max_price:
            max_price = price
            max_item = dp
    except (ValueError, KeyError):
        continue

print(f"Max price: {max_price}, Item: {max_item['title'] if max_item else 'N/A'}")
```
